```python
import rasterio
import numpy as np


def load_hyperspectral_image(file_path):

    with rasterio.open(file_path) as src:

        image = src.read()

        metadata = {
            "width": src.width,
            "height": src.height,
            "bands": src.count,
            "crs": str(src.crs),
            "transform": src.transform,
            "dtype": str(src.dtypes[0]),
        }

    print("Image shape:", image.shape)
    print("Metadata:", metadata)

    return image, metadata


image, metadata = load_hyperspectral_image(
    r"E:\internhips\infotact solutions\TerraSpectra Hyperspectral Crop Disease Forecasting\backend\data\TerraSpectra_Synthetic_Demo_Dataset\pixel_spectra.csv"
)
```
