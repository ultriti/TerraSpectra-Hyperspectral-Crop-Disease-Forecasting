```python
import numpy as np
from pathlib import Path

# ==========================================
# 1. LOAD HYPERSPECTRAL DATASET
# ==========================================

DATA_PATH = Path(
    r"E:\internhips\infotact solutions\TerraSpectra Hyperspectral Crop Disease Forecasting\backend\data\TerraSpectra_Synthetic_Demo_Dataset\synthetic_hsi_demo.npz"
)

data = np.load(DATA_PATH)

cube = data["cube"]
labels = data["labels"]

print("Hyperspectral Cube:", cube.shape)
print("Labels:", labels.shape)

# ==========================================
# 2. EXTRACT SPATIAL PATCHES
# ==========================================

PATCH_SIZE = 3
PAD = PATCH_SIZE // 2

height, width, bands = cube.shape

padded_cube = np.pad(cube, ((PAD, PAD), (PAD, PAD), (0, 0)), mode="reflect")

patches = []

for row in range(height):
    for col in range(width):

        patch = padded_cube[row : row + PATCH_SIZE, col : col + PATCH_SIZE, :]

        patches.append(patch)

patches = np.asarray(patches, dtype=np.float32)

# ==========================================
# 3. PREPARE LABELS
# ==========================================

y = labels.reshape(-1)

# ==========================================
# 4. VERIFY OUTPUT
# ==========================================

print("\nSpatial Feature Extraction Results")
print("----------------------------------")

print("Patch Dataset Shape:", patches.shape)
print("Label Dataset Shape:", y.shape)

print("Patch Size:", PATCH_SIZE)
print("Spectral Bands:", bands)
print("Features per Patch:", PATCH_SIZE * PATCH_SIZE * bands)

assert patches.shape == (height * width, 3, 3, bands)
assert len(patches) == len(y)

print("\nSpatial feature extraction completed!")

# ==========================================
# 5. SAVE EXTRACTED FEATURES
# ==========================================

OUTPUT_DIR = Path(
    r"E:\internhips\infotact solutions\TerraSpectra Hyperspectral Crop Disease Forecasting\backend\data\processed"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

np.savez_compressed(OUTPUT_DIR / "spatial_patches.npz", patches=patches, labels=y)

print("Saved:", OUTPUT_DIR / "spatial_patches.npz")
```
